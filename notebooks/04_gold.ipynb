```python
MY_ID = "preetam"
```

```python
# STEP 4 - Gold

gold = spark.table(f"workspace.capstone_{MY_ID}.silver_marks").join(
    spark.table(f"workspace.capstone_{MY_ID}.bronze_students").select("student_id","branch","section"),
    "student_id"
).join(
    spark.table(f"workspace.capstone_{MY_ID}.bronze_subjects"), "subject_code"
).select("student_id","branch","section","semester","subject_code","subject_name",
         "classes_held","classes_attended","attendance_pct","attendance_band",
         "internal_marks","external_marks","total_marks","sem_total_marks",
         "is_complete","sem_rank")

gold.write.mode("overwrite").saveAsTable(f"workspace.capstone_{MY_ID}.gold_student_subject_sem")
print(gold.count())   # expect 21,160
```
